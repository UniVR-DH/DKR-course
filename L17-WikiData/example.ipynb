```python
from SPARQLWrapper import SPARQLWrapper, JSON, XML, RDF
import pandas as pd
```

```python
sparql = SPARQLWrapper("https://query.wikidata.org/sparql") #determine SPARQL endpoint
sparql.setReturnFormat(JSON) #determine the output format
```

```python
#SPARQL query to be executed
sparql.setQuery("""
PREFIX wd:    <http://www.wikidata.org/entity/>
PREFIX wdt:   <http://www.wikidata.org/prop/direct/>
PREFIX rdfs:  <http://www.w3.org/2000/01/rdf-schema#>

SELECT ?author ?authorName ?birthPlace
WHERE {
  ?author wdt:P31 wd:Q36180 ;          # instance of writer
          rdfs:label ?authorName ;
          wdt:P19 ?birthPlace .       # place of birth
          ?birthPlace wdt:P31/wdt:P279* wd:Q515 . # place is an instance or a subclass of city

  FILTER (lang(?authorName) = "en")
}
ORDER BY ?authorName
""")

sparql.setReturnFormat(JSON)   # Return format is JSON
results = sparql.query().convert()   # execute SPARQL query and write result to "results"
```

```python
len(results['results']['bindings'])
```

```python
results_df = pd.json_normalize(results['results']['bindings'])
results_df[['author.value', 'authorName.value', 'birthPlace.value']]
```
